```python
import cv2
from pathlib import Path

pristine_dir = Path("./pristine_images_used_for_manipulation")
manip_dir = Path("./manipulated_images")

height_diffs = []
width_diffs = []

for manip_path in sorted(manip_dir.glob("*")):
    scene_id = manip_path.stem.rsplit("_", maxsplit=1)[0]

    pristine_path = pristine_dir / f"{scene_id}.jpg"

    pristine = cv2.imread(str(pristine_path))
    manipulated = cv2.imread(str(manip_path))

    if pristine is None or manipulated is None:
        continue

    height_diffs.append(pristine.shape[0] - manipulated.shape[0])
    width_diffs.append(pristine.shape[1] - manipulated.shape[1])

print("Unique height differences:")
print(sorted(set(height_diffs)))

print()

print("Unique width differences:")
print(sorted(set(width_diffs)))
```
