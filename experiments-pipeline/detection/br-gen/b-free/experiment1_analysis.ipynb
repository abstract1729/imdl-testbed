## Output Analysis

### Helper Functions

```python
def get_performance_subsets(df,n=4,threshold=0.5,):
    """
    Select best, worst, and borderline samples from an experiment DataFrame.

    Each returned dictionary has the form:

        {
            "real": [
                (image_path, score),
                ...
            ],
            "fake": [
                (image_path, score),
                ...
            ]
        }

    Parameters
    ----------
    df : pandas.DataFrame
        Experiment output DataFrame.

    n : int, default=4
        Number of samples to select per class.

    threshold : float, default=0.5
        Detection decision threshold.

    Returns
    -------
    tuple[dict, dict, dict]
        best, worst, borderline
    """

    required_columns = {
        "image_path",
        "label",
        "score",
        "inference_status",
    }

    missing = required_columns - set(df.columns)

    if missing:
        raise ValueError(
            f"Missing required columns: {sorted(missing)}"
        )

    # --------------------------------------------------------
    # Successful inferences only
    # --------------------------------------------------------

    df = df[df["inference_status"] == "SUCCESS"].copy()

    if df.empty:
        raise ValueError(
            "No successful inferences found."
        )

    # --------------------------------------------------------
    # Ensure numeric values
    # --------------------------------------------------------

    df["score"] = pd.to_numeric(
        df["score"],
        errors="coerce",
    )

    df["label"] = pd.to_numeric(
        df["label"],
        errors="coerce",
    )

    df = df.dropna(
        subset=["image_path", "score", "label"]
    )

    # --------------------------------------------------------
    # Split real / fake
    # --------------------------------------------------------

    real = df[df["label"] == 0].copy()
    fake = df[df["label"] == 1].copy()

    # --------------------------------------------------------
    # Best
    #
    # Real -> score closest to 0
    # Fake -> score closest to 1
    # --------------------------------------------------------

    best_real = (
        real.sort_values("score", ascending=True)
        .head(n)
    )

    best_fake = (
        fake.sort_values("score", ascending=False)
        .head(n)
    )

    best = {
        "real": list(
            zip(
                best_real["image_path"],
                best_real["score"],
            )
        ),
        "fake": list(
            zip(
                best_fake["image_path"],
                best_fake["score"],
            )
        ),
    }

    # --------------------------------------------------------
    # Worst
    #
    # Real -> score closest to 1
    # Fake -> score closest to 0
    # --------------------------------------------------------

    worst_real = (
        real.sort_values("score", ascending=False)
        .head(n)
    )

    worst_fake = (
        fake.sort_values("score", ascending=True)
        .head(n)
    )

    worst = {
        "real": list(
            zip(
                worst_real["image_path"],
                worst_real["score"],
            )
        ),
        "fake": list(
            zip(
                worst_fake["image_path"],
                worst_fake["score"],
            )
        ),
    }

    # --------------------------------------------------------
    # Borderline
    #
    # Closest to the decision threshold.
    # --------------------------------------------------------

    real["threshold_distance"] = (
        real["score"] - threshold
    ).abs()

    fake["threshold_distance"] = (
        fake["score"] - threshold
    ).abs()

    borderline_real = (
        real.sort_values(
            "threshold_distance",
            ascending=True,
        )
        .head(n)
    )

    borderline_fake = (
        fake.sort_values(
            "threshold_distance",
            ascending=True,
        )
        .head(n)
    )

    borderline = {
        "real": list(
            zip(
                borderline_real["image_path"],
                borderline_real["score"],
            )
        ),
        "fake": list(
            zip(
                borderline_fake["image_path"],
                borderline_fake["score"],
            )
        ),
    }

    return best, worst, borderline
```

```python
import matplotlib.pyplot as plt
from PIL import Image


def _format_score(score):
    """
    Format prediction score for visualization.

    Scores close to 0 or 1 are displayed in scientific notation
    to avoid hiding their actual values through rounding.
    """
    score = float(score)

    if score < 0.01 or score > 0.99:
        return f"{score:.3e}"

    return f"{score:.2f}"


def visualize_prediction_samples(samples):
    """
    Visualize real and fake samples in a 2x4 grid.

    Parameters
    ----------
    samples : dict
        Dictionary with the structure:

        {
            "real": [
                (image_path, score),
                ...
            ],
            "fake": [
                (image_path, score),
                ...
            ]
        }

    Layout
    ------
    Row 1 -> Real images
    Row 2 -> Fake images
    """

    fig, axes = plt.subplots(
        2,
        4,
        figsize=(16, 8)
    )

    row_labels = ["Real", "Fake"]

    for row, label in enumerate(["real", "fake"]):

        samples_for_class = samples.get(label, [])

        for col in range(4):

            ax = axes[row, col]

            # Handle missing samples
            if col >= len(samples_for_class):
                ax.axis("off")
                continue

            image_path, score = samples_for_class[col]

            try:
                image = Image.open(image_path).convert("RGB")

                ax.imshow(image)

                ax.set_title(
                    f"Score: {_format_score(score)}",
                    fontsize=12
                )

                ax.axis("off")

            except Exception as e:
                ax.axis("off")
                ax.set_title(
                    f"Error\n{type(e).__name__}",
                    fontsize=10
                )

        # Explicit row label
        axes[row, 0].annotate(
            row_labels[row],
            xy=(-0.12, 0.5),
            xycoords="axes fraction",
            fontsize=14,
            fontweight="bold",
            ha="right",
            va="center",
            rotation=90
        )

    # Leave space on the left for row labels
    plt.tight_layout(rect=[0.04, 0, 1, 1])

    plt.show()
```

## Threshold Calibration

```python
# ============================================================
# Threshold Calibration
# ============================================================
import numpy as np
from sklearn.metrics import (precision_score,recall_score,f1_score,
                             accuracy_score,balanced_accuracy_score,)

# ------------------------------------------------------------
# Ground truth and continuous B-Free scores
# ------------------------------------------------------------

y_true = successful_df["label"].astype(int).to_numpy()
y_score = successful_df["score"].astype(float).to_numpy()
```

```python
# ------------------------------------------------------------
# Sweep thresholds
# ------------------------------------------------------------

thresholds = np.linspace(0.0, 1.0, 1001)

threshold_results = []

for threshold in thresholds:

    y_pred = (y_score >= threshold).astype(int)
    precision = precision_score(y_true,y_pred,zero_division=0)
    recall = recall_score(y_true,y_pred,zero_division=0)
    f1 = f1_score(y_true,y_pred,zero_division=0)
    accuracy = accuracy_score(y_true,y_pred)
    balanced_accuracy = balanced_accuracy_score(y_true,y_pred)

    threshold_results.append({
        "Threshold": threshold,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "Accuracy": accuracy,
        "Balanced Accuracy": balanced_accuracy,
    })

threshold_df = pd.DataFrame(threshold_results)
```

```python
# ------------------------------------------------------------
# Find thresholds where BOTH Precision and Recall >= 90%
# ------------------------------------------------------------

valid_thresholds = threshold_df[ (threshold_df["Precision"] >= 0.90) & (threshold_df["Recall"] >= 0.90) ].copy()
print("=" * 70)
print(f"{DATASET} : THRESHOLD CALIBRATION")
print("=" * 70)
output_file = (PROJECT_ROOT/ 'outputs' / TASK/ DATASET/ MODEL/ f"{DATASET}_threshold_calibration.csv")

if not valid_thresholds.empty:
    # Among valid thresholds, select the one
    # with the highest F1 score.
    best = valid_thresholds.sort_values(by=["F1", "Balanced Accuracy"],ascending=False).iloc[0]
    
    print("\nThreshold achieving BOTH Precision >= 90% "
          "and Recall >= 90%:")

    print(f"\nBest Threshold     : {best['Threshold']:.4f}")
    print(f"Precision          : {best['Precision']:.4f}")
    print(f"Recall             : {best['Recall']:.4f}")
    print(f"F1                 : {best['F1']:.4f}")
    print(f"Accuracy           : {best['Accuracy']:.4f}")
    print(f"Balanced Accuracy  : {best['Balanced Accuracy']:.4f}")

    print(
        f"\nNumber of valid thresholds : "
        f"{len(valid_thresholds)}"
    )

    print(
        f"Valid threshold range     : "
        f"{valid_thresholds['Threshold'].min():.4f} "
        f"to "
        f"{valid_thresholds['Threshold'].max():.4f}"
    )

else:

    print(
        "\nNo threshold achieves BOTH "
        "Precision >= 90% and Recall >= 90%."
    )

    # Best F1 threshold as fallback
    best = threshold_df.sort_values(
        by=["F1", "Balanced Accuracy"],
        ascending=False
    ).iloc[0]

    print("\nBest available Precision/Recall tradeoff "
          "using maximum F1:")

    print(f"\nBest Threshold     : {best['Threshold']:.4f}")
    print(f"Precision          : {best['Precision']:.4f}")
    print(f"Recall             : {best['Recall']:.4f}")
    print(f"F1                 : {best['F1']:.4f}")
    print(f"Accuracy           : {best['Accuracy']:.4f}")
    print(f"Balanced Accuracy  : {best['Balanced Accuracy']:.4f}")


# ------------------------------------------------------------
# Save threshold sweep
# ------------------------------------------------------------

threshold_df.to_csv(
    output_file,
    index=False
)

print(
    f"\nFull threshold sweep saved to: "
    f"{DATASET}_threshold_calibration.csv"
)
```

## Saving Misclassifications

```python
from pathlib import Path
import math

import pandas as pd
from PIL import Image, ImageDraw, ImageFont


def save_classification_analysis(
    results_df,
    task,
    dataset,
    model,
    experiment,
    project_root,
    threshold=0.5,
    correct_percentile=0.70,
):
    """
    Save correctly classified and misclassified samples.

    Output structure
    ----------------
    outputs/
        <task>/
            <dataset>/
                <model>/
                    classification_analysis/
                        misclassifications/
                            real/
                            edited/
                            misclassifications.csv

                        correctly_classified/
                            real/
                            edited/
                            correctly_classified.csv

    Misclassifications
    ------------------
    False positives:
        GT = Real
        Pred = Edited
        Sorted by descending score.

    False negatives:
        GT = Edited
        Pred = Real
        Sorted by ascending score.

    Correct classifications
    -----------------------
    True negatives:
        GT = Real
        Pred = Real
        Sorted by ascending score.

    True positives:
        GT = Edited
        Pred = Edited
        Sorted by descending score.

    Only the first `correct_percentile` fraction of correctly
    classified samples in each class is saved.

    Each saved image contains a separate metadata header above
    the original image. The original image pixels are untouched.

    Parameters
    ----------
    results_df : pandas.DataFrame
        Experiment output DataFrame.

    task : str
        Task name.

    dataset : str
        Dataset name.

    model : str
        Model name.

    experiment : str
        Experiment name.

    project_root : str or Path
        Project root directory.

    threshold : float, default=0.5
        Classification threshold.

    correct_percentile : float, default=0.70
        Fraction of correctly classified samples to retain.

        Example:
            100 correct samples -> 70 saved
            250 correct samples -> 175 saved

    Returns
    -------
    tuple[pandas.DataFrame, pandas.DataFrame]
        (
            misclassification_df,
            correctly_classified_df
        )
    """

    project_root = Path(project_root).resolve()

    # ========================================================
    # Validate required columns
    # ========================================================

    required_columns = {
        "image_id",
        "file_name",
        "label",
        "image_path",
        "score",
        "inference_status",
    }

    missing_columns = required_columns - set(results_df.columns)

    if missing_columns:
        raise ValueError(
            "Missing required columns in results_df: "
            f"{sorted(missing_columns)}"
        )

    # ========================================================
    # Successful inferences only
    # ========================================================

    df = results_df[
        results_df["inference_status"]
        .astype(str)
        .str.upper()
        == "SUCCESS"
    ].copy()

    if df.empty:
        raise ValueError(
            "No successful inferences found in results_df."
        )

    # ========================================================
    # Numeric conversion
    # ========================================================

    df["label"] = pd.to_numeric(
        df["label"],
        errors="coerce",
    )

    df["score"] = pd.to_numeric(
        df["score"],
        errors="coerce",
    )

    df = df.dropna(
        subset=[
            "image_id",
            "image_path",
            "label",
            "score",
        ]
    )

    # ========================================================
    # Validate percentile
    # ========================================================

    if not 0 < correct_percentile <= 1:
        raise ValueError(
            "`correct_percentile` must be between 0 and 1."
        )

    # ========================================================
    # Generate predictions
    # ========================================================

    df["predicted_label"] = (
        df["score"] >= threshold
    ).astype(int)

    # ========================================================
    # Separate correctly classified / misclassified samples
    # ========================================================

    correctly_classified = df[
        df["label"] == df["predicted_label"]
    ].copy()

    false_positives = df[
        (df["label"] == 0)
        & (df["predicted_label"] == 1)
    ].copy()

    false_negatives = df[
        (df["label"] == 1)
        & (df["predicted_label"] == 0)
    ].copy()

    # ========================================================
    # MISCLASSIFICATION SORTING
    # ========================================================

    # False positives:
    # Real image predicted as Edited.
    #
    # Higher score = more confidently wrong.
    false_positives = (
        false_positives
        .sort_values(
            "score",
            ascending=False,
        )
        .reset_index(drop=True)
    )

    # False negatives:
    # Edited image predicted as Real.
    #
    # Lower score = more confidently wrong.
    false_negatives = (
        false_negatives
        .sort_values(
            "score",
            ascending=True,
        )
        .reset_index(drop=True)
    )

    # ========================================================
    # CORRECT CLASSIFICATION SORTING
    # ========================================================

    correctly_real = correctly_classified[
        correctly_classified["label"] == 0
    ].copy()

    correctly_fake = correctly_classified[
        correctly_classified["label"] == 1
    ].copy()

    # Real correctly classified:
    # Lower score = more confidently Real.
    correctly_real = (
        correctly_real
        .sort_values(
            "score",
            ascending=True,
        )
        .reset_index(drop=True)
    )

    # Edited correctly classified:
    # Higher score = more confidently Edited.
    correctly_fake = (
        correctly_fake
        .sort_values(
            "score",
            ascending=False,
        )
        .reset_index(drop=True)
    )

    # ========================================================
    # 70% CAP FOR CORRECTLY CLASSIFIED SAMPLES
    # ========================================================

    max_correct_real = math.ceil(
        len(correctly_real)
        * correct_percentile
    )

    max_correct_fake = math.ceil(
        len(correctly_fake)
        * correct_percentile
    )

    correctly_real = correctly_real.head(
        max_correct_real
    )

    correctly_fake = correctly_fake.head(
        max_correct_fake
    )

    # ========================================================
    # Output directories
    # ========================================================

    output_root = (
        project_root
        / "outputs"
        / task
        / dataset
        / model
        / "classification_analysis"
    )

    misclassification_root = (
        output_root / "misclassifications"
    )

    correct_root = (
        output_root / "correctly_classified"
    )

    misclassification_real_dir = (
        misclassification_root / "real"
    )

    misclassification_edited_dir = (
        misclassification_root / "edited"
    )

    correct_real_dir = (
        correct_root / "real"
    )

    correct_edited_dir = (
        correct_root / "edited"
    )

    for directory in [
        misclassification_real_dir,
        misclassification_edited_dir,
        correct_real_dir,
        correct_edited_dir,
    ]:
        directory.mkdir(
            parents=True,
            exist_ok=True,
        )

    # ========================================================
    # Font helpers
    # ========================================================

    def get_font(size):
        """
        Load a macOS font, with fallback to PIL default.
        """

        font_paths = [
            "/System/Library/Fonts/Helvetica.ttc",
            "/System/Library/Fonts/SFNS.ttf",
            "/Library/Fonts/Arial.ttf",
        ]

        for font_path in font_paths:
            try:
                return ImageFont.truetype(
                    font_path,
                    size,
                )
            except OSError:
                continue

        return ImageFont.load_default()

    def get_font_for_width(width):
        """
        Adapt font size according to image width.
        """

        if width >= 1200:
            size = 30
        elif width >= 800:
            size = 26
        elif width >= 600:
            size = 23
        elif width >= 400:
            size = 20
        elif width >= 300:
            size = 18
        else:
            size = 16

        return get_font(size)

    # ========================================================
    # Text wrapping
    # ========================================================

    def wrap_text_by_width(
        text,
        font,
        max_width,
        draw,
    ):
        """
        Wrap text according to rendered pixel width.
        """

        words = text.split()

        if not words:
            return [""]

        lines = []
        current_line = words[0]

        for word in words[1:]:

            candidate = (
                current_line
                + " "
                + word
            )

            bbox = draw.textbbox(
                (0, 0),
                candidate,
                font=font,
            )

            candidate_width = (
                bbox[2] - bbox[0]
            )

            if candidate_width <= max_width:
                current_line = candidate
            else:
                lines.append(current_line)
                current_line = word

        lines.append(current_line)

        # ----------------------------------------------------
        # Handle very long tokens such as image IDs.
        # ----------------------------------------------------

        final_lines = []

        for line in lines:

            bbox = draw.textbbox(
                (0, 0),
                line,
                font=font,
            )

            if (
                bbox[2] - bbox[0]
                <= max_width
            ):
                final_lines.append(line)
                continue

            current = ""

            for char in line:

                candidate = current + char

                bbox = draw.textbbox(
                    (0, 0),
                    candidate,
                    font=font,
                )

                if (
                    bbox[2] - bbox[0]
                    <= max_width
                ):
                    current = candidate
                else:

                    if current:
                        final_lines.append(
                            current
                        )

                    current = char

            if current:
                final_lines.append(current)

        return final_lines

    # ========================================================
    # Create annotated image
    # ========================================================

    def create_annotated_image(
        image_path,
        rank,
        image_id,
        true_label,
        predicted_label,
        score,
    ):
        """
        Create an image with a separate metadata header.

        The original image is never modified.
        """

        image_path = Path(image_path)

        image = Image.open(
            image_path
        ).convert("RGB")

        width, height = image.size

        font = get_font_for_width(
            width
        )

        # ----------------------------------------------------
        # Temporary drawing surface for text measurement
        # ----------------------------------------------------

        temp_canvas = Image.new(
            "RGB",
            (width, 100),
            "white",
        )

        temp_draw = ImageDraw.Draw(
            temp_canvas
        )

        horizontal_padding = 15

        max_text_width = (
            width
            - 2 * horizontal_padding
        )

        # ----------------------------------------------------
        # Score formatting
        # ----------------------------------------------------

        score = float(score)

        if score < 0.01 or score > 0.99:
            score_text = f"{score:.6e}"
        else:
            score_text = f"{score:.4f}"

        # ----------------------------------------------------
        # Metadata
        # ----------------------------------------------------

        rank_text = (
            f"Rank: {rank:06d}"
        )

        id_text = (
            f"ID: {image_id}"
        )

        gt_text = (
            "REAL"
            if true_label == 0
            else "EDITED"
        )

        pred_text = (
            "REAL"
            if predicted_label == 0
            else "EDITED"
        )

        prediction_text = (
            f"GT: {gt_text}    "
            f"Pred: {pred_text}    "
            f"Score: {score_text}"
        )

        # ----------------------------------------------------
        # Wrap image ID
        # ----------------------------------------------------

        id_lines = wrap_text_by_width(
            id_text,
            font,
            max_text_width,
            temp_draw,
        )

        # ----------------------------------------------------
        # Calculate dimensions
        # ----------------------------------------------------

        bbox = temp_draw.textbbox(
            (0, 0),
            "Ag",
            font=font,
        )

        line_height = (
            bbox[3] - bbox[1]
        )

        line_spacing = 8
        vertical_padding = 15

        total_lines = (
            1
            + len(id_lines)
            + 1
        )

        header_height = (
            (2 * vertical_padding)
            + (total_lines * line_height)
            + (
                (total_lines - 1)
                * line_spacing
            )
        )

        # ----------------------------------------------------
        # Final canvas
        # ----------------------------------------------------

        canvas = Image.new(
            "RGB",
            (
                width,
                height + header_height,
            ),
            "white",
        )

        draw = ImageDraw.Draw(
            canvas
        )

        y = vertical_padding

        # Rank
        draw.text(
            (
                horizontal_padding,
                y,
            ),
            rank_text,
            fill="black",
            font=font,
        )

        y += (
            line_height
            + line_spacing
        )

        # Image ID
        for line in id_lines:

            draw.text(
                (
                    horizontal_padding,
                    y,
                ),
                line,
                fill="black",
                font=font,
            )

            y += (
                line_height
                + line_spacing
            )

        # GT / Prediction / Score
        draw.text(
            (
                horizontal_padding,
                y,
            ),
            prediction_text,
            fill="black",
            font=font,
        )

        # ----------------------------------------------------
        # Paste original image underneath header
        # ----------------------------------------------------

        canvas.paste(
            image,
            (
                0,
                header_height,
            ),
        )

        return canvas

    # ========================================================
    # Save samples
    # ========================================================

    def save_samples(
        dataframe,
        output_directory,
        classification_name,
    ):
        """
        Save images and return manifest rows.
        """

        manifest_rows = []

        for index, row in dataframe.iterrows():

            rank = index + 1

            source_path = Path(
                row["image_path"]
            )

            if not source_path.exists():

                print(
                    "Warning: image not found, "
                    f"skipping:\n{source_path}"
                )

                continue

            annotated_image = (
                create_annotated_image(
                    image_path=source_path,
                    rank=rank,
                    image_id=row["image_id"],
                    true_label=int(
                        row["label"]
                    ),
                    predicted_label=int(
                        row["predicted_label"]
                    ),
                    score=float(
                        row["score"]
                    ),
                )
            )

            # ------------------------------------------------
            # Preserve original extension
            # ------------------------------------------------

            extension = (
                source_path.suffix.lower()
            )

            if not extension:
                extension = ".jpg"

            destination = (
                output_directory
                / f"{rank:06d}{extension}"
            )

            annotated_image.save(
                destination
            )

            manifest_rows.append(
                {
                    "rank": rank,
                    "image_id": row["image_id"],
                    "file_name": row["file_name"],
                    "true_label": int(
                        row["label"]
                    ),
                    "predicted_label": int(
                        row["predicted_label"]
                    ),
                    "score": float(
                        row["score"]
                    ),
                    "threshold": threshold,
                    "distance_from_threshold": abs(
                        float(row["score"])
                        - threshold
                    ),
                    "classification": (
                        classification_name
                    ),
                    "original_image_path": str(
                        source_path
                    ),
                    "saved_image_path": str(
                        destination
                    ),
                }
            )

        return manifest_rows

    # ========================================================
    # Save misclassifications
    # ========================================================

    misclassification_rows = []

    misclassification_rows.extend(
        save_samples(
            dataframe=false_positives,
            output_directory=(
                misclassification_real_dir
            ),
            classification_name=(
                "false_positive"
            ),
        )
    )

    misclassification_rows.extend(
        save_samples(
            dataframe=false_negatives,
            output_directory=(
                misclassification_edited_dir
            ),
            classification_name=(
                "false_negative"
            ),
        )
    )

    # ========================================================
    # Save correctly classified samples
    # ========================================================

    correct_rows = []

    correct_rows.extend(
        save_samples(
            dataframe=correctly_real,
            output_directory=(
                correct_real_dir
            ),
            classification_name=(
                "true_negative"
            ),
        )
    )

    correct_rows.extend(
        save_samples(
            dataframe=correctly_fake,
            output_directory=(
                correct_edited_dir
            ),
            classification_name=(
                "true_positive"
            ),
        )
    )

    # ========================================================
    # Create manifest DataFrames
    # ========================================================

    misclassification_df = pd.DataFrame(
        misclassification_rows
    )

    correctly_classified_df = pd.DataFrame(
        correct_rows
    )

    # ========================================================
    # Save manifests
    # ========================================================

    misclassification_manifest = (
        misclassification_root
        / "misclassifications.csv"
    )

    correctly_classified_manifest = (
        correct_root
        / "correctly_classified.csv"
    )

    misclassification_df.to_csv(
        misclassification_manifest,
        index=False,
    )

    correctly_classified_df.to_csv(
        correctly_classified_manifest,
        index=False,
    )

    # ========================================================
    # Summary
    # ========================================================

    print("=" * 70)
    print("Classification Analysis")
    print("=" * 70)

    print(f"Experiment : {experiment}")
    print(f"Dataset    : {dataset}")
    print(f"Model      : {model}")
    print(f"Threshold  : {threshold}")
    print()

    print("Misclassifications")
    print(
        f"  False Positives : "
        f"{len(false_positives)}"
    )
    print(
        f"  False Negatives : "
        f"{len(false_negatives)}"
    )

    print()

    print("Correctly Classified")
    print(
        f"  Real retained   : "
        f"{len(correctly_real)}"
    )
    print(
        f"  Edited retained : "
        f"{len(correctly_fake)}"
    )

    print()

    print(
        f"Output directory : "
        f"{output_root}"
    )

    print(
        f"Misclassification manifest : "
        f"{misclassification_manifest}"
    )

    print(
        f"Correct classification manifest : "
        f"{correctly_classified_manifest}"
    )

    return (
        misclassification_df,
        correctly_classified_df,
    )
```

## Misclassification Analysis (for BR-Gen)

```python
import pandas as pd


def analyze_brgen_misclassifications(
    results_df,
    threshold=0.5,
):
    """
    Analyze BR-GEN misclassifications across its primary groups.

    BR-GEN image ID conventions
    ----------------------------

    Real images:
        real_<dataset>_<image_id>

    Example:
        real_imagenet_12345
        real_coco_67890
        real_places_54321

    Fake images:
        fake_<edit_model>_<stuff/background>_<dataset>_<image_id>

    Example:
        fake_lama_stuff_imagenet_12345
        fake_mat_background_coco_67890

    Analysis is performed independently for:

        1. Editing model
        2. Edit type
        3. Source dataset

    For real images, edit_model and edit_type are set to:

        "N/A (real)"

    This is intentional because those attributes do not exist
    for real images.

    For each grouping, the function reports:

        - false_positive
        - false_negative
        - total_misclassified

    Parameters
    ----------
    results_df : pandas.DataFrame
        Experiment output DataFrame.

    threshold : float, default=0.5
        Threshold used to convert scores into predictions.

    Returns
    -------
    dict
        {
            "by_edit_model": DataFrame,
            "by_edit_type": DataFrame,
            "by_source_dataset": DataFrame
        }
    """

    # ========================================================
    # Validate required columns
    # ========================================================

    required_columns = {
        "image_id",
        "label",
        "score",
        "inference_status",
    }

    missing = required_columns - set(results_df.columns)

    if missing:
        raise ValueError(
            f"Missing required columns: {sorted(missing)}"
        )

    # ========================================================
    # Successful inferences only
    # ========================================================

    df = results_df[
        results_df["inference_status"]
        .astype(str)
        .str.upper()
        == "SUCCESS"
    ].copy()

    if df.empty:
        raise ValueError(
            "No successful inferences found."
        )

    # ========================================================
    # Numeric conversion
    # ========================================================

    df["label"] = pd.to_numeric(
        df["label"],
        errors="coerce",
    )

    df["score"] = pd.to_numeric(
        df["score"],
        errors="coerce",
    )

    df = df.dropna(
        subset=[
            "image_id",
            "label",
            "score",
        ]
    )

    # ========================================================
    # Generate predictions
    # ========================================================

    df["predicted_label"] = (
        df["score"] >= threshold
    ).astype(int)

    # ========================================================
    # Parse BR-GEN IDs
    # ========================================================

    valid_edit_types = {
        "stuff",
        "background",
    }

    valid_datasets = {
        "imagenet",
        "coco",
        "places",
    }

    def parse_image_id(image_id):

        parts = str(image_id).split("_")

        if len(parts) < 2:
            return pd.Series(
                {
                    "edit_model": None,
                    "edit_type": None,
                    "source_dataset": None,
                }
            )

        image_type = parts[0].lower()

        # ====================================================
        # REAL
        #
        # real_<dataset>_<image_id>
        # ====================================================

        if image_type == "real":

            if len(parts) < 3:
                return pd.Series(
                    {
                        "edit_model": None,
                        "edit_type": None,
                        "source_dataset": None,
                    }
                )

            source_dataset = parts[1].lower()

            if source_dataset not in valid_datasets:
                return pd.Series(
                    {
                        "edit_model": None,
                        "edit_type": None,
                        "source_dataset": None,
                    }
                )

            return pd.Series(
                {
                    # These are intentionally N/A because
                    # real images have no editing model/type.
                    "edit_model": "N/A (real)",
                    "edit_type": "N/A (real)",
                    "source_dataset": source_dataset,
                }
            )

        # ====================================================
        # FAKE
        #
        # fake_<edit_model>_<edit_type>_<dataset>_<image_id>
        # ====================================================

        if image_type == "fake":

            if len(parts) < 5:
                return pd.Series(
                    {
                        "edit_model": None,
                        "edit_type": None,
                        "source_dataset": None,
                    }
                )

            # ------------------------------------------------
            # Find stuff/background
            # ------------------------------------------------

            edit_type_index = None

            for i, part in enumerate(parts):

                if part.lower() in valid_edit_types:
                    edit_type_index = i
                    break

            if edit_type_index is None:
                return pd.Series(
                    {
                        "edit_model": None,
                        "edit_type": None,
                        "source_dataset": None,
                    }
                )

            # ------------------------------------------------
            # Find dataset after edit type
            # ------------------------------------------------

            dataset_index = None

            for i in range(
                edit_type_index + 1,
                len(parts),
            ):

                if parts[i].lower() in valid_datasets:
                    dataset_index = i
                    break

            if dataset_index is None:
                return pd.Series(
                    {
                        "edit_model": None,
                        "edit_type": None,
                        "source_dataset": None,
                    }
                )

            # ------------------------------------------------
            # Everything between "fake" and edit type
            # constitutes the editing model.
            # ------------------------------------------------

            edit_model = "_".join(
                parts[1:edit_type_index]
            )

            if not edit_model:
                return pd.Series(
                    {
                        "edit_model": None,
                        "edit_type": None,
                        "source_dataset": None,
                    }
                )

            return pd.Series(
                {
                    "edit_model": edit_model.lower(),
                    "edit_type": parts[
                        edit_type_index
                    ].lower(),
                    "source_dataset": parts[
                        dataset_index
                    ].lower(),
                }
            )

        # ====================================================
        # Unknown image type
        # ====================================================

        return pd.Series(
            {
                "edit_model": None,
                "edit_type": None,
                "source_dataset": None,
            }
        )

    # ========================================================
    # Apply parser
    # ========================================================

    parsed = df["image_id"].apply(
        parse_image_id
    )

    df = pd.concat(
        [
            df,
            parsed,
        ],
        axis=1,
    )

    # ========================================================
    # Validate parsing
    #
    # Only genuinely unparseable rows are removed.
    # "N/A (real)" is a VALID value.
    # ========================================================

    invalid_rows = df[
        df[
            [
                "edit_model",
                "edit_type",
                "source_dataset",
            ]
        ].isna().any(axis=1)
    ]

    if not invalid_rows.empty:

        print(
            f"Warning: {len(invalid_rows)} image IDs "
            "could not be parsed."
        )

        print(
            "Examples:"
        )

        print(
            invalid_rows["image_id"]
            .head(10)
            .tolist()
        )

        df = df.drop(
            invalid_rows.index
        )

    # ========================================================
    # Identify misclassifications
    # ========================================================

    df["misclassification"] = "correct"

    # False positive:
    # Real -> Edited
    df.loc[
        (df["label"] == 0)
        & (df["predicted_label"] == 1),
        "misclassification",
    ] = "false_positive"

    # False negative:
    # Edited -> Real
    df.loc[
        (df["label"] == 1)
        & (df["predicted_label"] == 0),
        "misclassification",
    ] = "false_negative"

    misclassified = df[
        df["misclassification"] != "correct"
    ].copy()

    # ========================================================
    # Generic grouping function
    # ========================================================

    def create_group_table(
        dataframe,
        group_column,
    ):
        """
        Create FP/FN/total count table.

        N/A (real) is retained as an explicit group so that
        false positives are not silently dropped.
        """

        if dataframe.empty:

            return pd.DataFrame(
                columns=[
                    group_column,
                    "false_positive",
                    "false_negative",
                    "total_misclassified",
                ]
            )

        table = pd.crosstab(
            dataframe[group_column],
            dataframe["misclassification"],
        )

        # Ensure both columns always exist.
        for column in [
            "false_positive",
            "false_negative",
        ]:

            if column not in table.columns:
                table[column] = 0

        table = table[
            [
                "false_positive",
                "false_negative",
            ]
        ]

        table["total_misclassified"] = (
            table["false_positive"]
            + table["false_negative"]
        )

        table = (
            table
            .reset_index()
            .sort_values(
                "total_misclassified",
                ascending=False,
            )
            .reset_index(drop=True)
        )

        return table

    # ========================================================
    # Generate analyses
    # ========================================================

    by_edit_model = create_group_table(
        misclassified,
        "edit_model",
    )

    by_edit_type = create_group_table(
        misclassified,
        "edit_type",
    )

    by_source_dataset = create_group_table(
        misclassified,
        "source_dataset",
    )

    # ========================================================
    # Overall summary
    # ========================================================

    total_samples = len(df)

    total_misclassified = len(
        misclassified
    )

    total_fp = len(
        misclassified[
            misclassified["misclassification"]
            == "false_positive"
        ]
    )

    total_fn = len(
        misclassified[
            misclassified["misclassification"]
            == "false_negative"
        ]
    )

    # ========================================================
    # Print results
    # ========================================================

    print("=" * 70)
    print("BR-GEN Misclassification Analysis")
    print("=" * 70)

    print(
        f"Successful samples       : "
        f"{total_samples}"
    )

    print(
        f"Total misclassifications : "
        f"{total_misclassified}"
    )

    print(
        f"False positives          : "
        f"{total_fp}"
    )

    print(
        f"False negatives          : "
        f"{total_fn}"
    )

    print()

    print(
        "Misclassifications by Editing Model"
    )

    display(by_edit_model)

    print(
        "Misclassifications by Edit Type"
    )

    display(by_edit_type)

    print(
        "Misclassifications by Source Dataset"
    )

    display(by_source_dataset)

    # ========================================================
    # Return
    # ========================================================

    return {
        "by_edit_model": by_edit_model,
        "by_edit_type": by_edit_type,
        "by_source_dataset": by_source_dataset,
    }
```

## Testing the Sampling strategy

```python
from pathlib import Path
import pandas as pd


def load_subset_ids(subset_file):
    """
    Load image IDs from a benchmark subset TXT file.

    Lines beginning with '#' and blank lines are ignored.

    Parameters
    ----------
    subset_file : str or Path
        Path to the subset TXT file.

    Returns
    -------
    list[str]
        Image IDs contained in the subset.
    """

    subset_file = Path(subset_file)

    if not subset_file.exists():
        raise FileNotFoundError(
            f"Subset file not found:\n{subset_file}"
        )

    image_ids = []

    with subset_file.open("r") as f:

        for line in f:

            line = line.strip()

            # Ignore metadata/header/comment lines
            if not line or line.startswith("#"):
                continue

            image_ids.append(line)

    return image_ids


def parse_brgen_id(image_id):
    """
    Parse a BR-GEN image ID.

    Real:
        real_<dataset>_<image_id>

    Fake:
        fake_<edit_model>_<edit_type>_<dataset>_<image_id>

    Returns
    -------
    dict
        Parsed grouping information.
    """

    parts = str(image_id).split("_")

    if len(parts) < 3:
        raise ValueError(
            f"Invalid BR-GEN image ID: {image_id}"
        )

    image_type = parts[0].lower()

    valid_datasets = {
        "imagenet",
        "coco",
        "places",
    }

    valid_edit_types = {
        "stuff",
        "background",
    }

    # ========================================================
    # REAL
    #
    # real_<dataset>_<image_id>
    # ========================================================

    if image_type == "real":

        source_dataset = parts[1].lower()

        if source_dataset not in valid_datasets:
            raise ValueError(
                f"Unknown source dataset in ID: {image_id}"
            )

        return {
            "image_id": image_id,
            "image_type": "real",
            "edit_model": "N/A (real)",
            "edit_type": "N/A (real)",
            "source_dataset": source_dataset,
        }

    # ========================================================
    # FAKE
    #
    # fake_<edit_model>_<edit_type>_<dataset>_<image_id>
    # ========================================================

    if image_type == "fake":

        # Find stuff/background rather than assuming that
        # edit_model is always a single token.
        edit_type_index = None

        for i, part in enumerate(parts):

            if part.lower() in valid_edit_types:
                edit_type_index = i
                break

        if edit_type_index is None:
            raise ValueError(
                f"Could not identify edit type in ID: {image_id}"
            )

        # Dataset occurs after edit type.
        dataset_index = None

        for i in range(
            edit_type_index + 1,
            len(parts),
        ):

            if parts[i].lower() in valid_datasets:
                dataset_index = i
                break

        if dataset_index is None:
            raise ValueError(
                f"Could not identify source dataset in ID: {image_id}"
            )

        edit_model = "_".join(
            parts[1:edit_type_index]
        )

        edit_type = parts[
            edit_type_index
        ].lower()

        source_dataset = parts[
            dataset_index
        ].lower()

        if not edit_model:
            raise ValueError(
                f"Could not identify edit model in ID: {image_id}"
            )

        return {
            "image_id": image_id,
            "image_type": "fake",
            "edit_model": edit_model.lower(),
            "edit_type": edit_type,
            "source_dataset": source_dataset,
        }

    raise ValueError(
        f"Unknown image type in BR-GEN ID: {image_id}"
    )
```

```python
def analyze_brgen_subset_distribution(
    real_subset_file,
    fake_subset_file,
):
    """
    Analyze the BR-GEN test subset distribution.

    The function parses both the real and fake subset TXT files
    and creates a dataframe containing the grouping variables:

        - image_type
        - edit_model
        - edit_type
        - source_dataset

    Real images:
        edit_model = "N/A (real)"
        edit_type  = "N/A (real)"

    Fake images:
        edit_model = actual editing model
        edit_type  = stuff/background
        source_dataset = imagenet/coco/places

    Parameters
    ----------
    real_subset_file : str or Path
        Real subset TXT file.

    fake_subset_file : str or Path
        Fake subset TXT file.

    Returns
    -------
    pd.DataFrame
        One row per image.
    """

    # ========================================================
    # Load IDs
    # ========================================================

    real_ids = load_subset_ids(
        real_subset_file
    )

    fake_ids = load_subset_ids(
        fake_subset_file
    )

    # ========================================================
    # Parse IDs
    # ========================================================

    records = []

    # Real
    for image_id in real_ids:

        records.append(
            parse_brgen_id(image_id)
        )

    # Fake
    for image_id in fake_ids:

        records.append(
            parse_brgen_id(image_id)
        )

    # ========================================================
    # Create dataframe
    # ========================================================

    distribution_df = pd.DataFrame(
        records
    )

    # ========================================================
    # Basic sanity checks
    # ========================================================

    if len(distribution_df) != (
        len(real_ids) + len(fake_ids)
    ):
        raise RuntimeError(
            "Number of parsed records does not match "
            "number of subset IDs."
        )

    return distribution_df
```
