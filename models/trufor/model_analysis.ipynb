```python
import torch
from pathlib import Path
from pprint import pprint
```

```python
ROOT = Path.cwd()

CHECKPOINT_PATH = ROOT / "TruFor_train_test"/ "pretrained_models" / "trufor.pth.tar"

print(CHECKPOINT_PATH)

ckpt = torch.load(CHECKPOINT_PATH,map_location="cpu",weights_only=False)
print("=" * 60)
print("TOP LEVEL KEYS")
print("=" * 60)

print(list(ckpt.keys()))
```

```python
print("=" * 60)
print("METADATA")
print("=" * 60)

for k, v in ckpt.items():

    if isinstance(v, (int, float, str, bool)):
        print(f"{k:25s}: {v}")

    elif isinstance(v, dict):
        print(f"{k:25s}: dict ({len(v)})")

    elif isinstance(v, list):
        print(f"{k:25s}: list ({len(v)})")

    elif torch.is_tensor(v):
        print(f"{k:25s}: Tensor {tuple(v.shape)}")

    else:
        print(f"{k:25s}: {type(v)}")
```

```python
state_dict = ckpt["state_dict"]
print(f"Total tensors : {len(state_dict)}")
```

```python
for i, key in enumerate(state_dict.keys()):
    print(key)

    # if i == 80:
    #     break
```
